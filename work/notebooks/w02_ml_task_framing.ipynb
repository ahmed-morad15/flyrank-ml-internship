## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My lane as an ML task: Ranking / Scoring

I frame this lane as a ranking/scoring problem because the decision is not simply whether a page is good or bad. The goal is to prioritize which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring. The model would produce a priority score for each content page, allowing the content or SEO team to review the highest-priority pages first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target or proxy

The initial target will be a future, observed content outcome rather than a hand-written rule. For the starter dataset, I will use the observed change in search performance as a proxy for page priority and investigate whether a later time-window outcome can be defined consistently. The current `trend_direction` field is useful for exploration, but I will not treat a rule-derived label as proof of future performance.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success metric

The primary success metric will be Precision@K, starting with Precision@50. This measures how many of the top 50 ranked pages are relevant review candidates. This fits the real decision because the content team has limited review capacity and needs a useful shortlist rather than a prediction for every page.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

### Unit of analysis

The unit of analysis is one content page (content item). Each row represents one page and contains its client, search, content, engagement, and trend-related signals. The ranking output will therefore assign a priority score to each content page.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML beats a fixed rule here

A fixed rule can prioritize pages using one or a small number of manually chosen thresholds, but it may miss interactions between search, content, engagement, visibility, and trend signals. A ranking/scoring model can combine multiple observed signals and produce a consistent priority score across pages.

The Week 1 starter experiment provides an initial indication that learned ranking can outperform a simple hand-written rule: Random Forest achieved a Precision@50 of 0.740 compared with 0.240 for the hand-written baseline on the starter evaluation. This is evidence that a learned approach may be useful for prioritization, not proof that ML will always outperform rules or that a recommended content action will improve performance.
