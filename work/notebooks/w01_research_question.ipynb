## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### The question

**Research question:** Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

**Decision:** Decide which content pages deserve limited review time first.

**Who acts:** A content or SEO reviewer.

**Action:** The reviewer can inspect the highest-priority pages and decide whether to refresh, expand, protect, prune, or monitor them based on the available evidence.

**Cost of a wrong call:** A false positive can waste limited content-team time on a page that does not need attention. A false negative can cause a potentially valuable page to be overlooked.

**Why data or ML can help:** The dataset contains multiple observable signals about search visibility, traffic, content characteristics, engagement, and trends. Combining these signals may help rank review candidates more consistently than relying on one rule or metric alone.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

# Load the starter dataset
path = "data/raw/content_refresh_anonymized.csv"

if os.path.exists(path):
    df = pd.read_csv(path)
else:
    url = "https://raw.githubusercontent.com/ahmed-morad15/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
    df = pd.read_csv(url)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")

# Number and share of pages currently trending down
down_pages = (df["trend_direction"] == "down").sum()
down_pct = down_pages / len(df) * 100

print(f"Pages trending down: {down_pages:,} ({down_pct:.1f}%)")

# Median recent impressions
median_impressions = df["impressions_90d"].median()
print(f"Median impressions_90d: {median_impressions:,.0f}")

# Median average search position
median_position = df["avg_position"].median()
print(f"Median avg_position: {median_position:.2f}")
```
